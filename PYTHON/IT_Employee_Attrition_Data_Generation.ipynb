```python
import pandas as pd
import numpy as np

# Setting a seed so the random data remains the same every time
np.random.seed(42)
total_records = 1000

# Generating basic column values
emp_id = np.arange(1001, 1001 + total_records)
dept = np.random.choice(['IT', 'Sales', 'HR', 'Operations'], size=total_records, p=[0.45, 0.25, 0.10, 0.20])
exp_years = np.random.randint(1, 13, size=total_records)
promotion_delay = np.random.randint(0, 6, size=total_records)
work_mode = np.random.choice(['WFH', 'Hybrid', 'WFO'], size=total_records, p=[0.30, 0.40, 0.30])
overtime = np.random.choice(['Yes', 'No'], size=total_records, p=[0.35, 0.65])
commute_distance = np.random.randint(2, 45, size=total_records)
```

```python
# Logic: Salary calculation based on experience
salary_lpa = []

for exp in exp_years:
    # Base formula: (Experience * 1.4) + some random allowance between 2.5 to 4.5
    calculated_salary = (exp * 1.4) + np.random.uniform(2.5, 4.5)
    salary_lpa.append(round(calculated_salary, 2))
```

```python
attrition_list = []

for i in range(total_records):
    # Starting with a very low base probability of 5%
    leave_probability = 0.05

    # Pillar 1: Salary & Promotion impact
    if salary_lpa[i] < 6.0:
        leave_probability += 0.15
    if promotion_delay[i] >= 3:
        leave_probability += 0.10

    # Pillar 2: Work-Life Balance impact
    if overtime[i] == 'Yes':
        leave_probability += 0.20
    if commute_distance[i] > 25 and work_mode[i] == 'WFO':
        leave_probability += 0.12

    # Pillar 3: Experience level impact (Freshers change jobs faster)
    if exp_years[i] <= 2:
        leave_probability += 0.15

    # Pillar 4 & 5: Department & Work Mode impact
    if dept[i] == 'Sales':
        leave_probability += 0.05
    if work_mode[i] == 'WFO':
        leave_probability += 0.05

    # Final check: If random number is less than probability, they leave
    if np.random.rand() < leave_probability:
        attrition_list.append('Yes')
    else:
        attrition_list.append('No')
```

```python
# Combining everything into a final DataFrame
df = pd.DataFrame({
    'Employee_ID': emp_id,
    'Department': dept,
    'Total_Experience_Years': exp_years,
    'Salary_LPA': salary_lpa,
    'Years_Since_Last_Promotion': promotion_delay,
    'Work_Mode': work_mode,
    'Distance_From_Home_KM': commute_distance,
    'Overtime': overtime,
    'Attrition': attrition_list
})
```

```python
# Saving the data to a CSV file
df.to_csv('it_employee_attrition_data.csv', index=False)
print("Data file created successfully!")
```
